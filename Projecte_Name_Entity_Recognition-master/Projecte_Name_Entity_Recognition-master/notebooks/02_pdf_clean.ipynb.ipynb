```python
import sys
sys.path.append('..')

from ner.pdf import extract_pdf_dataset
from ner.inference import load_gliner_model, predict_dataset   # o ner.inference si ya lo renombraste
from ner.utils import load_jsonl, save_responses_to_jsonl
from ner.evaluation import evaluate_ner_strict, evaluate_ner_relaxed

# Los datos correctos (gold).
GOLD_PATH = '../data/plain/validation.jsonl'
# Carpeta con los PDF
PDF_DIR = '../data/pdf_clean'
# Donde guardamos las predicciones
OUTPUT_PATH = '../results/pdf_clean/gliner_small_validation.jsonl'

LABELS = [
    'PERSON_NAME', 'ID_CARD_NUMBER', 'ADDRESS',
    'REGISTRATION_NUMBER', 'DATE', 'AMOUNT', 'ORGANIZATION'
]
```

```python
# Leemos los registros y buscamos el PDF de cada uno (PDF_DIR/<id>.pdf)
records = load_jsonl(GOLD_PATH)
samples = extract_pdf_dataset(records, PDF_DIR)

print(len(samples))
print(samples[0]["text"][:500])   # comprobamos que el texto se ve bien
```

```python
model = load_gliner_model()
responses = predict_dataset(model, samples, LABELS, threshold=0.5)
save_responses_to_jsonl(responses, OUTPUT_PATH)
```

```python
strict_results = evaluate_ner_strict(GOLD_PATH, OUTPUT_PATH)
relaxed_results = evaluate_ner_relaxed(GOLD_PATH, OUTPUT_PATH)
```
